```python

s = input("Enter a string: ")

vowels = "aeiouAEIOU"
v = [ch for ch in s if ch in vowels]

v.reverse()

result = ""
i = 0

for ch in s:
    if ch in vowels:
        result += v[i]
        i += 1
    else:
        result += ch

print("String after reversing vowels:", result)
```

```python
#original(mam notes)
def reverse_vowels(s):
    vowels = "aeiouAEIOU"
    chars = list(s)
    left, right = 0, len(chars) - 1

    while left < right:
        if chars[left] not in vowels:
            left += 1
        elif chars[right] not in vowels:
            right -= 1
        else:
            chars[left], chars[right] = chars[right], chars[left]
            left += 1
            right -= 1

    return "".join(chars)

print(reverse_vowels("hello"))
```
